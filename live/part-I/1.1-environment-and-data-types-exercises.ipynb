## Exercise 2: Unit conversions

A temperature is recorded as 18.5 °C. Compute it in kelvin and in degrees Fahrenheit. Separately, a discharge of 4.7 m³ s⁻¹ drains a catchment of area 250 km². Express the specific discharge in mm per day (use: 1 m³ s⁻¹ spread over 1 km² for one day equals 86.4 mm). Print each result to two decimals with an f-string.

````python
# Temperature
temperature_c = 18.5

temperature_k = temperature_c + 273.15
temperature_f = temperature_c * 9/5 + 32

# Specific discharge
discharge_m3_s = 4.7
catchment_area_km2 = 250

specific_discharge_mm_day = discharge_m3_s / catchment_area_km2 * 86.4

# Print results
print(f"Temperature: {temperature_k:.2f} K")
print(f"Temperature: {temperature_f:.2f} °F")
print(f"Specific discharge: {specific_discharge_mm_day:.2f} mm/day")
````

## Exercise 3: Operator families and a quality flag

Three daily readings arrive from a station: `reading_1_celsius` = 18.2, `reading_2_celsius` = -300.0, and `reading_3_celsius` = None (the sensor reported nothing).

1. Build a boolean `is_plausible` that is `True` when the first reading is above absolute zero and below 60 °C.
2. Build `is_implausible` for the second reading: apply the same test as in step 1, and negate it with `not`.
3. Use `is` to check whether the third reading is missing.
4. Starting from `n_valid = 0`, use `+=` twice to count how many of the first two readings pass the step 1 test. Note that `int()` casts `True` to 1 and `False` to 0.
5. The sensor is rated for -40 °C to 60 °C, both bounds included. Use `>=` and `<=` to test whether the first reading lies in that range, and `!=` to check that the first two readings differ.

````python
reading_1_celsius = 18.2
reading_2_celsius = -300.0
reading_3_celsius = None

# 1. Check whether the first reading is plausible
is_plausible = reading_1_celsius > -273.15 and reading_1_celsius < 60

# 2. Check whether the second reading is implausible
is_implausible = not (reading_2_celsius > -273.15 and reading_2_celsius < 60)

# 3. Check whether the third reading is missing
is_missing = reading_3_celsius is None

# 4. Count valid readings using +=
n_valid = 0
n_valid += int(reading_1_celsius > -273.15 and reading_1_celsius < 60)
n_valid += int(reading_2_celsius > -273.15 and reading_2_celsius < 60)

# 5. Check sensor range and whether readings differ
is_in_sensor_range = reading_1_celsius >= -40 and reading_1_celsius <= 60
readings_differ = reading_1_celsius != reading_2_celsius

print(f"is_plausible: {is_plausible}")
print(f"is_implausible: {is_implausible}")
print(f"is_missing: {is_missing}")
print(f"n_valid: {n_valid}")
print(f"is_in_sensor_range: {is_in_sensor_range}")
print(f"readings_differ: {readings_differ}")
````

## Exercise 5: Indexing and slicing a record

Weather records often arrive in a fixed shape, where each field always occupies the same
positions. Given

```python

```

the first three characters are the station code and the rest is a date as `dd.mm.yyyy`.

1. Using **only indexing and slicing** (no `.split()`), extract the station code, the day, the
   month, and the four-digit year, and print them.
2. Now extract the month a second way, using `.split(".")` on the date part, and use `==` to

3. In a comment, say which of the two approaches you would trust to extract the month from a
   record where the station code can be two or four characters long, and why.

````python
record = "ALO01.03.2022"

# 1. Using indexing and slicing
station_code = record[:3]
day = record[3:5]
month = record[6:8]
year = record[9:13]

print(station_code)
print(day)
print(month)
print(year)

# 2. Extract the month using split()
date_part = record[3:]
month_split = date_part.split(".")[1]

print(month_split)
print(month == month_split)

# 3. For variable-length station codes, I would trust split()
# because fixed positions would change if the station code has
# two or four characters, whereas split(".") identifies the
# month based on the separators.
````

## Exercise 7: Round-trip through a file, with an append

Write the Jungfraujoch week from the lecture — the daily maximum temperatures -8.4, -8.9, -9.2,
-9.0, -9.5 (°C) — to `_files/jfj_week.txt`, one value per line, using `pathlib` and a `with`
block. Create the `_files` folder first if it does not exist; do not rely on Exercise 6 having
made it.

1. Write the five values with mode `"w"`, one `.write()` call per value.
2. Reopen the file with mode `"a"` and append the last two days of the week, `-7.8` and `-6.1`.
3. Read the file back with `.readlines()`, cast each of the seven lines to `float` by name, and
   compute the mean by summing the seven values by hand with `+` and dividing by `len()`. Print
   it to two decimals.

````python
from pathlib import Path

# Create the "_files" folder if it does not exist
files_folder = Path("_files")
files_folder.mkdir(exist_ok=True)

# Define the path of the file
file_path = files_folder / "jfj_week.txt"

# Daily maximum temperatures for the first five days
temp_1 = -8.4
temp_2 = -8.9
temp_3 = -9.2
temp_4 = -9.0
temp_5 = -9.5

# Open the file in "w" mode and write one value per line
with file_path.open("w") as file:
    file.write(f"{temp_1}\n")
    file.write(f"{temp_2}\n")
    file.write(f"{temp_3}\n")
    file.write(f"{temp_4}\n")
    file.write(f"{temp_5}\n")

# Open the same file in "a" mode and append the last two days
with file_path.open("a") as file:
    file.write("-7.8\n")
    file.write("-6.1\n")

# Read all seven lines from the file
with file_path.open("r") as file:
    lines = file.readlines()

# Convert each line from text (string) to a float
value_1 = float(lines[0])
value_2 = float(lines[1])
value_3 = float(lines[2])
value_4 = float(lines[3])
value_5 = float(lines[4])
value_6 = float(lines[5])
value_7 = float(lines[6])

# Calculate the mean by adding the seven values and dividing by their number
mean_temperature = (
    value_1 + value_2 + value_3 + value_4 + value_5 + value_6 + value_7
) / len(lines)

# Print the mean with two decimal places
print(f"Mean temperature: {mean_temperature:.2f} °C")
````

### Exercise 8: Same seven values, same mean?

Adding the same floats in a different order does not have to give exactly the same result,
because each `+` rounds its result to the nearest float. This exercise uses `math.isclose` from
the going-deeper box on floating-point precision.

1. Compute the mean of the Jungfraujoch week — -8.4, -8.9, -9.2, -9.0, -9.5, -7.8, -6.1 — by
   adding the values in that order and dividing by 7.
2. Compute it again, adding the same seven values in reverse order.
3. Compare the two means with `==`, and print the result.
4. Compare the same two means with `math.isclose`, and print the result.
5. In a comment, say which of the two comparisons you would put in a script that checks whether
   two calculations of the same quantity agree, and why.

````python
import math

# The seven daily maximum temperatures
temp_1 = -8.4
temp_2 = -8.9
temp_3 = -9.2
temp_4 = -9.0
temp_5 = -9.5
temp_6 = -7.8
temp_7 = -6.1

# 1. Calculate the mean in the original order
mean_forward = (
    temp_1 + temp_2 + temp_3 + temp_4 + temp_5 + temp_6 + temp_7
) / 7

# 2. Calculate the mean in reverse order
mean_reverse = (
    temp_7 + temp_6 + temp_5 + temp_4 + temp_3 + temp_2 + temp_1
) / 7

# 3. Compare the two means using ==
print(f"Using ==: {mean_forward == mean_reverse}")

# 4. Compare the two means using math.isclose()
print(f"Using math.isclose(): {math.isclose(mean_forward, mean_reverse)}")

# 5. In a scientific script, I would use math.isclose()
# because tiny floating-point differences can occur even when
# two calculations represent the same quantity.
````
